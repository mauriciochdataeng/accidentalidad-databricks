# 01_Generacion_o_Ingesta

Lectura inicial de los archivos CSV ubicados en el Unity Catalog Volume del proyecto.

En esta primera etapa se valida la lectura del archivo `accidentes.csv` antes de crear la tabla Bronze.

```python
# Parámetros del proyecto
dbutils.widgets.text("catalog_name", "accidentalidad_databricks", "Catalog del proyecto")

PROY_CATALOG = dbutils.widgets.get("catalog_name")
PROY_SCHEMA_BRONZE = "bronze"
RAW_PATH = f"/Volumes/{PROY_CATALOG}/{PROY_SCHEMA_BRONZE}/raw_files"

print(f"RAW_PATH: {RAW_PATH}")
```

## Lectura inicial de accidentes

Se usa `header=True` para tomar la primera fila como nombres de columnas y `inferSchema=False` para conservar inicialmente los valores como texto en Bronze.

```python
df_accidentes = (
    spark.read
    .option("header", True)
    .option("inferSchema", False)
    .csv(f"{RAW_PATH}/accidentes.csv")
)
```

## Validación visual

Antes de escribir una tabla Delta, se revisan las primeras filas para confirmar que el CSV fue interpretado correctamente.

```python
display(df_accidentes.limit(10))
```
